```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import seaborn as sns
```

```python
df1=df.copy()
df2=df.copy()
df3=df.copy()
df4=df.copy()
df5=df.copy()
df6=df.copy()
df7=df.copy()
df8=df.copy()
df9=df.copy()
```

```python
#Label Encoding on the Area column
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df["Area"]=le.fit_transform(df["Area"])
df.head()
```

```python
#Perform Label  Encoding using loop
for col in df1.columns:
    if df1[col].dtype == np.number:
        continue
    else:
        df1[col]=le.fit_transform(df1[col])
```

```python
df1.head()
```

```python
from pandas.core.dtypes.common import is_numeric_dtype
```

```python
for c in  df2.columns:
    if is_numeric_dtype(df2[c]):
        continue
    else:
        df2[c]=le.fit_transform(df2[c])

df2.head()
    
```

```python
X=df2.drop(['Profit'],axis=1)
X.shape
```

```python
Y=df2[['Profit']]
Y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(X,Y,random_state=42,test_size=0.30)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(xtrain,ytrain)
```

```python
ypred = lr.predict(xtest)
print("R² score:", r2_score(ytest, ypred))
print("MSE:", mean_squared_error(ytest, ypred))
print("MAE:", mean_absolute_error(ytest, ypred))
```
