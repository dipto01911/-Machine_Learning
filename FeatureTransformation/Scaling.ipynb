```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
df=pd.read_csv('supershops.csv')
```

```python
df.head()
df1=df.copy()
df2=df.copy()
df3=df.copy()
df4=df.copy()
```

Perform linear regression without any encoding and Normalizationb

```python
x=df.drop(['Area','Profit'],axis=1)
y=df[['Profit']]
```

```python
x.head()
```

```python
y.shape
```

```python
xtrain1,xtest1,ytrain1,ytest1=train_test_split(x,y,random_state=20,test_size=0.30)
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
model=LinearRegression()
```

```python
model.fit(xtrain1,ytrain1)
```

```python
ypred = model.predict(xtest1)
print("R² score:", r2_score(ytest1, ypred))
print("MSE:", mean_squared_error(ytest1, ypred))
print("MAE:", mean_absolute_error(ytest1, ypred))
```

Normalization

```python
from sklearn.preprocessing import MinMaxScaler
m=MinMaxScaler()
```

```python
df_mar=m.fit(df[['Marketing Spend']])
df_mar
```

```python
df[["Marketing Spend"]]=m.transform(df[['Marketing Spend']])
df.head()
```

```python
#Scaling all numeric row except target 

scaler = MinMaxScaler()

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    if col != "Profit":   # target বাদ
        df[[col]] = scaler.fit_transform(df[[col]])

df.head()
```

```python
X=df.drop('Profit',axis=1)
Y=df[['Profit']]
```

```python
Y.shape
```

```python
X.shape
```

```python
encoder = OneHotEncoder(drop='first', sparse_output=False)
cat_cols = X.select_dtypes(include='object').columns
for col in cat_cols:
    
    encoded = encoder.fit_transform(X[[col]])
    
    encoded_df = pd.DataFrame(
        encoded,
        columns=encoder.get_feature_names_out([col])
    )
    
    X = pd.concat([X, encoded_df], axis=1)
    df.drop(col, axis=1, inplace=True)

X.head()
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(X,Y,random_state=30,test_size=0.30)
```

```python
print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
lr=LinearRegression()
lr.fit(xtrain,ytrain)
```

```python
ypred = lr.predict(xtest)
print("R² score:", r2_score(ytest, ypred))
print("MSE:", mean_squared_error(ytest, ypred))
print("MAE:", mean_absolute_error(ytest, ypred))
```
