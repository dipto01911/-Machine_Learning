Text Vectorizer

```python
import pandas as pd
import numpy as np
df=pd.read_csv('data.csv')
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

Count Vectorizer

```python
cv=CountVectorizer()
```

```python
cv_x=cv.fit_transform(df['test'])
cv_x
```

```python
cv_x.toarray()
```

```python
cv.get_feature_names_out()
```

```python
cv_df=pd.DataFrame(cv_x.toarray(),columns=cv.get_feature_names_out(),index=df['test'])
cv_df
```

TfidfVectorizer(tf=term_frequency and idf=inverse_document_frequency)

```python
tf_z=tf.fit_transform(df['test'])
tf_z
```
